## DATA QUALITY FINDINGS AND CLEANING DECISIONS


The initial inspection identified several data quality issues in the raw sales dataset.

| Issue  |                               | Finding  |                                 | Planned action  |

|---|---|---|
                            
| Duplicate records |                    | 5 duplicate rows identified |              |Remove confirmed exact duplictes from the cleaned copy|
| Missing values |                       | 9 missing values across columns |          |Handle each column according to the available evidence|
| Inconsistent text |                    | Trailing space in Region and Product_category| |Remove unnecessary leading and trailing space|
| Date format |                          |3 dates require special handling |        |Standardize date carefully without misinterpreting formats|
| Missing Unit_price |                   | One Printer price is missing |              |Fill with 180,supported by the other Printer records|
| Missing Cost |                         | One cost value is missing |                 |Keep missing until a reliable value can be established| |  Invalid looking quantity |              |One record has Quantity=0 but Sales=450|     |Flag for review;do not change automatically|
| Sales calculation inconsistency|  |One record has a 25% discount but recorded Sales does not match| |Flag for review;don't alter without evidence

# Data Protection
The original DataFrame ,df, will remain unchanged.All cleanig and transformation will be performed on a separate copy named df_clean.

Unresolved issues will be documented rather than silently corrected.This preserves data integrity and makes the analysis reproducible.



```python
#CREATE A CLEAN WORKING COPY
#ORIGINAL df STAYS UNCHANGED
df_clean = df.copy()
```

```python
#Remove duplicate rows from the working copy while preserving original dataset
df_clean = df_clean.drop_duplicates()
```

```python
#Verify the duplicate were removed
df_clean.duplicated().sum()
```

```python
#Clean extra spaces in text
#This removes unnecessary spaces from Region and Product_Category
df_clean["Region"] = df_clean["Region"].str.strip()
df_clean["Product_Category"] = df_clean["Product_Category"].str.strip()
```

```python
#Fix the missing Printer price
#We can safely use 180 because we already confirmed that all other Printer records have that unit price
df_clean.loc[df_clean["Product"] == "Printer", "Unit_Price"] = df_clean.loc[df_clean["Product"] == "Printer","Unit_Price"].fillna(180)
```

```python
#Handle Missing Value 
df_clean.isna().sum()
```

```python
#Standardize the Order Dates
#This converts  the dates to a consistent datetime format while preserving the intended month first interpretation
df_clean["Order_Date"] = pd.to_datetime(df_clean["Order_Date"],format="mixed",dayfirst=False,errors="coerce")
```

```python
#Checking whether any date has failed to convert
df_clean["Order_Date"].isna().sum()
```

```python
#Calculate Profit
#Adding a Profit column
#The row with missing Cost will have missing Profit
df_clean["Profit"] = df_clean["Sales"] - df_clean["Cost"]
```

```python
#Check Profit Results
df_clean[["Order_ID","Sales","Cost","Profit"]].head(10)
```

```python
#Check for negative Profits
df_clean[df_clean["Profit"] < 0][["Order_ID","Sales","Cost","Profit"]]
```

```python
#Cleaned Dataset
df_clean.to_csv("../data/sales_data_cleaned.csv",index=False)
```

```python
#Sales Visualisation
#Regional sales totals for our bar chart
region_sales = df_clean.groupby("Region")["Sales"].sum().sort_values(ascending=False)
```

```python
#Add exact Sales values above each bar
ax = region_sales.plot(kind="bar",title = "Total Sales by Region")

ax.set_xlabel("Region")
ax.set_ylabel("Total Sales")
ax.bar_label(ax.containers[0], fmt="{:,.0f}")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Sales by Product
#This arranges Product from highest to lowest Sales
product_sales = df_clean.groupby("Product")["Sales"].sum().sort_values(ascending=False)
```

```python
#Sales_by_Product BarChart
ax = product_sales.plot(kind="bar",figsize=(10,5))

ax.set_title("Total Sales by Product")
ax.set_xlabel("Product")
ax.set_ylabel("Total Sales")
ax.bar_label(ax.containers[0],fmt="{:,.0f}",rotation=90)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
#prepare monthly sales
monthly_sales = df_clean.groupby(df_clean["Order_Date"].dt.month)["Sales"].sum()
```

```python
#Monthly Sales Line Chart
ax = monthly_sales.plot(kind="line",marker = "o",figsize=(9,5))

ax.set_title("Monthly Sales Trend")
ax.set_xlabel("Month")
ax.set_ylabel("Total Sales")
ax.set_xticks(range(1,13))
ax.set_xticklabels([
    "Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"
])
for month , sales in monthly_sales.items():
    ax.annotate(
        f"{sales:,.0f}",
        (month,sales),
        textcoords="offset points",
        xytext=(0,8),
        ha="center",
        fontsize=8
    )
ax.set_ylim(bottom=0)
plt.tight_layout()
plt.show()
```

```python
#Profit by Product Chart
#This chart ranks product from highest to lowest total profit

product_profit = (df_clean.groupby("Product")["Profit"].sum().sort_values(ascending=False))

ax = product_profit.plot(kind = "bar",figsize=(10,5))

ax.set_title("Total Profit by Product")
ax.set_xlabel("Product")
ax.set_ylabel("Total Profit")

ax.bar_label(ax.containers[0],fmt="{:,.0f}",rotation=90)

plt.xticks(rotation=45,ha="right")
plt.tight_layout()
plt.show()
```

```python
#Verify Profit Margin
#This calculates the profit margin using only orders with known costs,so the sales and profit  figures are comparable

known_cost = df_clean["Cost"].notna()

profit_margin = (
    df_clean.loc[known_cost,"Profit"].sum() 
    / df_clean.loc[known_cost,"Sales"].sum()
)*100

print(f"Profit margin on orders with known costs: {profit_margin:.2f}%")
```

```python
#Update sales_summary_csv with the verified profit margin

known_cost = df_clean["Cost"].notna()

summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Cost",
        "Total Profit (known Costs)",
        "Profit Margin (known Costs)(%)"
    ],
    "Value": [
        df_clean["Sales"].sum(),
        df_clean["Cost"].sum(),
        df_clean["Profit"].sum(),
        (
            df_clean.loc[known_cost,"Profit"].sum()
            / df_clean.loc[known_cost,"Sales"].sum()
        )*100
    ]
})

summary.to_csv("../data/sales_summary.csv",index=False)

summary
```
